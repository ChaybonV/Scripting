### Oefening

Download logbook.txt van canvas, lees het en groepeer activiteiten per maand

--> 2021-03-09 13:13:40 - Started work on project E

output:

- January : 70 activities 
- February : 75 activities 
- March : 88 activities 
- April : 72 activities 
- ...

```python
file = open("logbook.txt", "r")
lines = file.readlines()
file.close()

months = ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"]
activity_count = {}

for line in lines:
    parts = line.split(" - ")
    if len(parts) > 1:
        if len(parts[0].split(" ")[0].split("-")) == 3:
            month = parts[0].split(" ")[0].split("-")[1]
            month_name = months[int(month) - 1]
            if month_name in activity_count:
                activity_count[month_name] += 1
            else:
                activity_count[month_name] = 1

for month in months:
    if month in activity_count:
        print("- " + month + " : " + str(activity_count[month]) + " activities")
```
