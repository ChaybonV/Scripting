Je hebt de volgende string: " Python programmeren is leuk en handig! Leer nu: slicen, indexen, trimmen en splitten. "

Gebruik string manipulatie technieken om de volgende resultaten te bereiken:

- Verwijder alle witruimte aan het begin en einde van de string.
- Extraheer het woord "Python" met index notatie.
- Extraheer de woorden "leuk en handig" met slicing notatie.

- Tel het aantal keer dat de letter 'e' voorkomt in de tekst.

```python
text = " Python programmeren is leuk en handig! Leer nu: slicen, indexen, trimmen en splitten. "

# 1. Verwijder alle witruimte aan het begin en einde van de string.
text = text.strip()
print(text)

# 2. Extraheer het woord "Python" met index notatie.
print(text[0:6])

# 3. Extraheer de woorden "leuk en handig" met slicing notatie.
print(text[23:37])

# 4. Keer de gehele string om (van achter naar voren).
print(text[::-1])

# 5. Maak een lijst van alle woorden in de tekst (gebruik een split methode).
print(text.split())

# 6. Extraheer alle woorden die beginnen met de letter 'p' (hoofdletter ongevoelig). 
print( [word for word in text.split() if word.lower().startswith('p')] )

# 7. Extraheer alleen de tweede zin uit de tekst (hint: split op '!').
print(text.split('!')[1])

# 8. Vervang alle spaties in de tekst door een underscore (_).
print(text.replace(' ', '_'))

# 9. Tel het aantal keer dat de letter 'e' voorkomt in de tekst.
print(text.lower().count('e'))

# 10. Creëer een nieuwe string met alleen de eerste letter van elk woord (hint: gebruik split en join).
print(''.join(word[0] for word in text.split()))
```

Implementeer een eenvoudige Caesarversleuteling waarbij je gebruik maakt van ASCII-waarden en isX-functies. De Caesarversleuteling is een van de oudste encryptietechnieken waarbij elke letter in een tekst wordt vervangen door een letter die een vast aantal posities verder in het alfabet staat. Je moet twee functies maken:

- versleutel(tekst, verschuiving) - Versleutelt de tekst met de gegeven verschuiving
- ontsleutel(tekst, verschuiving) - Ontsleutelt de versleutelde tekst terug naar de originele tekst

Specifieke vereisten:

- Alleen letters worden versleuteld; cijfers, spaties en leestekens blijven ongewijzigd
- Kleine letters blijven kleine letters, hoofdletters blijven hoofdletters
- De verschuiving moet "rondgaan" in het alfabet (als je voorbij 'z' gaat, begin je weer bij 'a')
- Gebruik ord() en chr() voor ASCII-manipulatie
- Gebruik isX-functies zoals islower(), isupper(), isalpha() om te bepalen welke tekens je moet versleutelen

```python
def versleutel(tekst, verschuiving):
    versleutelde_tekst = ""
    
    for char in tekst:
        if char.isalpha():
            nieuwe_char = chr(ord(char) + verschuiving)
            if (char.islower() and nieuwe_char > 'z') or (char.isupper() and nieuwe_char > 'Z'):
                nieuwe_char = chr(ord(char) - (26 - verschuiving))
            versleutelde_tekst += nieuwe_char
        else:
            versleutelde_tekst += char
            
    return versleutelde_tekst

def ontsleutel(tekst, verschuiving):
    ontsleutelde_tekst = ""
    
    for char in tekst:
        if char.isalpha():
            nieuwe_char = chr(ord(char) - verschuiving)
            if (char.islower() and nieuwe_char < 'a') or (char.isupper() and nieuwe_char < 'A'):
                nieuwe_char = chr(ord(char) + (26 - verschuiving))
            ontsleutelde_tekst += nieuwe_char
        else:
            ontsleutelde_tekst += char
            
    return ontsleutelde_tekst

tekst = "Hello World!"
verschuiving = 3
print("Versleuteld:",versleutel(tekst, verschuiving))
print("Ontsleuteld:",ontsleutel(versleutel(tekst, verschuiving), verschuiving))
```

## Regex

https://regexr.com/

- dan pas coderen!

```python
import re

text = """Dit is een regex test met woorden en getallen zoals 1 3 6,
        ah nee dit zijn cijfers ik bedoel ook echte getallen zoals 1567 en 89"""

# eerste getal
result = re.search(r"\d",text)
print(result.group()) # group ==> resultaat
print(result.span()) # span ==> locatie

# alle cijfers
result = re.findall(r"\d",text)
print(result)

# alle getallen
result = re.findall(r"\d+",text)
print(result)

regex_object = re.compile(r"\d")
text_split = regex_object.split(text)
print(text_split)

text_replace = regex_object.sub("DIGIT",text)
print(text_replace)
```

## Oefening

Je hebt een tekstbestand met contactgegevens in verschillende formaten. Gebruik regex-operaties om de gegevens te analyseren, te verwerken en op te schonen.

- Jan Janssen - tel: 06-12345678 - email: jan.janssen@example.com
- Petra de Vries (petra.devries@mail.nl) [Telefoonnummer: 0612345679]
- CONTACT: Mark Leenders | 06 87654321 | m.leenders@ergens.org
- Anne_Visser -- 0611223344 -- a.visser21@provider.com
- Karel Groot, email: kgroot@bedrijf.nl, Telefoon: 06-11112222

Doe nu het volgende;

- extraheer alle telefoonnummers
- verwerk nu alle contacten, een contact per regel, extraheer alle namen uit de contacten
- standaardiseer nu alle telefoonnummer volgens het formaat "06-XXXXXXXX" (met streepje)
- schrijf nu alles weg in een gestructureerde lijst van dictionaries met { naam: ***, e-mail: ***, telnr: *** }

```python
import re

text = """Jan Janssen - tel: 06-12345678 - email: jan.janssen@example.com
Petra de Vries (petra.devries@mail.nl) [Telefoonnummer: 0612345679]
CONTACT: Mark Leenders | 06 87654321 | m.leenders@ergens.org
Anne_Visser -- 0611223344 -- a.visser21@provider.com
Karel Groot, email: kgroot@bedrijf.nl, Telefoon: 06-11112222"""

phone_regex = r"(?:tel:|Telefoonnummer:|Telefoon:|--| |)06[\s\-]?\d{8}"
email_regex = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b"
name_regex = r"([A-Za-z]+(?:\s[A-Za-z]+)*)(?=\s*(?:-|--|,|\||\(|\d|\s))"

phones = re.findall(phone_regex, text)
emails = re.findall(email_regex, text)
names = re.findall(name_regex, text)

def standardize_phone_number(phone):
    phone = re.sub(r"[^0-9]", "", phone)
    return f"06-{phone[2:]}"  # Zet het telefoonnummer om naar 06-XXXXXXXX

# Controleer of de lengtes van de lijsten gelijk zijn
if len(names) == len(emails) == len(phones):
    contacts = []
    for i in range(len(names)):
        contacts.append({"naam": names[i], "e-mail": emails[i], "telnr": standardize_phone_number(phones[i])})
    
    for contact in contacts:
        print(contact)
else:
    print("Error: De lijsten van namen, e-mails en telefoonnummers zijn niet gelijk van lengte.")
    print(f"Namen: {len(names)}, E-mails: {len(emails)}, Telefoonnummers: {len(phones)}")
```
