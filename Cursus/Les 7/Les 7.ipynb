```python
import requests

headers = {
    "x-rapidapi-key": "a69cdae89amshdb1eb1b1b44f30bp1f4e96jsn8f47560ad780",
    "x-rapidapi-host": "spotify23.p.rapidapi.com"
}

def zoek_playlists(artiest_naam):
    url = "https://spotify23.p.rapidapi.com/search/"
    querystring = {"q": artiest_naam, "type": "playlists"}
    response = requests.get(url, headers=headers, params=querystring)

    data = response.json()
    playlists = []
    for item in data["playlists"]["items"]:
        playlist_data = item["data"]
        name = playlist_data["name"]
        url = playlist_data["uri"]
        playlists.append((name, f"https://open.spotify.com/playlist/{url.split(':')[2]}"))
    return playlists

artiest = input("Geef een artiestennaam op: ")
playlists = zoek_playlists(artiest)

if playlists:
    print("\nPlaylists die bij deze artiest passen:")
    for naam, link in playlists:
        print(f"- {naam}: {link}")
else:
    print("Geen relevante playlists gevonden.")
```
