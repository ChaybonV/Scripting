### Herhalingsoefening 

Maak een excel overzicht van alle boeken op deze pagina 

https://books.toscrape.com

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

URL = "https://books.toscrape.com/catalogue/"

books = []
page = 1

while True:
    url = f"{URL}page-{page}.html"
    response = requests.get(url)
    soup = BeautifulSoup(response.text)
    articles = soup.find_all("article", class_="product_pod")

    if not articles:
        break

    for article in articles:
        title = article.h3.a["title"]
        price = article.find("p", class_="price_color").text.strip()[2:]
        books.append([title, price])

    page += 1

df = pd.DataFrame(books, columns=["Title", "Price"])
df.to_excel("books.xlsx")

print("Excel file 'books.xlsx' is created!")
```

## Multithreading

Schrijf een programma dat de volgende stappen uitvoert:

- Maak twee functies function1() en function2().
   - function1() moet de getallen 1 tot en met 10 afdrukken, met een vertraging van 0,5 seconde tussen elk getal.
   - function2() moet de letters 'A' tot en met 'J' afdrukken, met een vertraging van 0,5 seconde tussen elke letter.
- Maak twee threads, één voor elke functie, en start ze tegelijkertijd.
- Wacht tot beide threads klaar zijn voordat u het programma verlaat.

```python
import time
import threading

def f1():
    for i in range(1, 11): # van 1 tot 10
        print(i)
        time.sleep(0.5)

def f2():
    for k in "ABCDEFGHIJ":
        print(k)
        time.sleep(0.5)

def main():
    thread1 = threading.Thread(target=f1)
    thread2 = threading.Thread(target=f2)

    thread1.start()
    thread2.start()

if __name__ == "__main__":
    main()
```

## Oefening

Maak een Python-programma dat webscraping gebruikt om gegevens te verzamelen van de Ultratop-website (https://www.ultratop.be/nl/annual.asp?year="2022"). De jaarparameter kan variëren tussen 2000 en 2024. Je programma moet de data voor elk jaar ophalen en opslaan in tekstbestanden.

Om het webscrapingproces te versnellen, implementeer je multi-threading in uw programma. Elke thread moet  een specifiek jaar ophalen en opslaan in een apart tekstbestand.

Zorg ervoor dat je programma nooit crasht!

```python
import requests
from bs4 import BeautifulSoup
import threading
import time
import os

def fetch_data(year):
    url = f'https://www.ultratop.be/nl/annual.asp?year={year}&cat=s'
    response = requests.get(url)    
    soup = BeautifulSoup(response.content)
    
    chart_data = soup.find_all('div', {"class":"content"})
        
    file_name = f'ultratop_{year}.txt'
    with open(file_name, 'w', encoding='utf-8') as file:
        for entry in chart_data:
            title_artist = entry.find('div', {"class":"chart_title"})
            if title_artist:
                song = title_artist.get_text(strip=True)
                file.write(f'{song}\n')
    
    print(f'Gegevens voor {year} zijn opgeslagen in {file_name}')

def fetch_for_years(start_year, end_year):
    threads = []
    for year in range(start_year, end_year + 1):
        thread = threading.Thread(target=fetch_data, args=(year,))
        threads.append(thread)
        thread.start()
        time.sleep(1)

    for thread in threads:
        thread.join()

if not os.path.exists('ultratop_data'):
    os.makedirs('ultratop_data')

os.chdir('ultratop_data')

fetch_for_years(2000, 2024)
```
