## Oefeningen 

1. Schrijf een functie die controleert of een getal een priemgetal is.

2. Gebruik daarna deze funcie in je zoektocht naar het volgende priemgetal na een ingegeven getal!

```python
# Oefening 1
def is_prime(number):
    if number <= 1:
        return False
    for i in range(2, number//2):
        if number % i == 0:
            return False
    return True

# Oefening 2
getal = int(input("Voer een getal in: "))
print(getal,"is priemgetal =",is_prime(getal))
getal+=1
while not is_prime(getal):
    getal+=1
print(getal,"is het volgende priemgetal")
```

## Oefeningen

1. Reken het gemiddelde uit per student. Kan je dit ook per toets doen? Voorbeeld: [[8, 7, 6, 9], [9, 9, 8, 8], [6, 7, 8, 7]]
2. Bereken de volgende zaken: alle getallen samen (unie), getallen die in allebei zitten (overeenkomst) en getallen die alleen in de eerste set zitten (verschil) Voorbeeld: {1, 2, 3, 4} en {3, 4, 5, 6}
3. Zet woorden in een woordenboek, waarbij de eerste letter de sleutel is en de woorden die met die letter beginnen eronder staan. Voorbeeld: ["appel", "boom", "aap", "banaan", "citroen", "aardbei"] → {'a': [...], 'b': [...], 'c': [...]}

```python
# Oefening 1
scores = [[8, 7, 6, 9], [9, 9, 8, 8], [6, 7, 8, 7]]

for student in scores:
    som = 0
    for toets in student:
        som+=toets
    print("Gemiddelde per student = ",som/len(student))

for i in range(0,len(scores[0])):
    som = 0
    for j in range(0,len(scores)):
        som += scores[j][i]
    print("Gemiddelde per toets =", som / len(scores))
```

```python
# Oefening 2
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

unie = set1 | set2
overeenkomst = set1 & set2
verschil = set1 - set2

print("Unie:", unie)
print("Overeenkomst:", overeenkomst)
print("Verschil:", verschil)
```

```python
# Oefening 3
woorden = ["appel", "boom", "aap", "banaan", "citroen", "aardbei"]
woorden_dict = {}

for woord in woorden:
    eerste_letter = woord[0]
    if eerste_letter in woorden_dict:
        woorden_dict[eerste_letter].append(woord)
    else:
        woorden_dict[eerste_letter] = [woord]

print(woorden_dict)
```

### Slicen

[start:einde / aantal:stap]

```python
import numpy as np

lijst = [x for x in range(0,20,2)]
print(lijst)

array = np.arange(10)*2
print(array)

print(lijst[0])
print(array[0])

print(lijst[0:3])
print(array[0:3])

print(lijst[0:-1:2])
print(array[0:-1:2])

print(array[-1:0])

print(lijst[::-1]) # --> Reverse van de lijst

print(array < 10)
print(np.where(array<10))
```

```python
def count_numbers():
    numbers = input("Give a list with numbers (seperated by spaces)").split()
    count_dict = {}
    for num in numbers:
        num = int(num)
        count_dict[num] = count_dict.get(num, 0) + 1
    return count_dict

result = count_numbers()
print(result)
```
