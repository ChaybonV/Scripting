## Oefening herhaling

Je script moet het password analyseren op basis van deze criteria:

Lengte:
- Minder dan 6 tekens: zeer zwak
- 6-7 tekens: zwak
- 8-10 tekens: redelijk
- 11-15 tekens: sterk
- Meer dan 15 tekens: zeer sterk

Karakterdiversiteit:
- Hoofdletters (A-Z)
- Kleine letters (a-z)
- Cijfers (0-9)
- Speciale tekens (!@#$%^&*()_+-=[]{}|;:'",.<>?/)

```python
import re

def analyse_password(password):    
    if len(password) < 6:
        sterkte = "Zeer zwak"
    elif len(password) < 8:
        sterkte = "Zwak"
    elif len(password) < 11:
        sterkte = "Redelijk"
    elif len(password) < 16:
        sterkte = "Sterk"
    else:
        sterkte = "Zeer sterk"

    heeft_grote_letter = bool(re.search(r'[A-Z]', password))
    heeft_kleine_letter = bool(re.search(r'[a-z]', password))
    heeft_cijfer = bool(re.search(r'\d', password))
    heeft_speciaal = bool(re.search(r'[!@#$%^&*()_+\-=\[\]{}|;:\'",.<>?/]', password))

    diversiteit = sum([heeft_grote_letter, heeft_kleine_letter, heeft_cijfer, heeft_speciaal])
    return f"Lengte: {sterkte}, Diversiteit: {diversiteit}/4 categorieën"

password = input("Geef een wachtwoord")
print(analyse_password(password))
```

Lees bestand dictianory in en haal alle palindromes er uit

```python
def is_palindrome(word):
    return word == word[::-1]

def find_palindromes(file_path):
    palindromes = []
    file = open(file_path, 'r')
    words = file.read().splitlines()
    for word in words:
        if is_palindrome(word):
            palindromes.append(word)
    file.close()
    return palindromes 

print(find_palindromes("dictionary.txt"))
```

## Countries
- Filter de dataset om alleen landen weer te geven met een bevolking van meer dan 50 miljoen.
- Sorteer de dataset op basis van de oppervlakte van de landen, van groot naar klein.
- Toon de top 10 landen met de hoogste bevolkingsdichtheid (bevolking per vierkante kilometer)
- Bereken de gemiddelde bevolking van alle landen in de dataset.
- Groepeer de landen per continent en bereken het totaal aantal inwoners per continent.
- Wat is het kleinste en grootste land qua oppervlakte in de dataset?
- Exporteer alleen de landen uit Europa naar een apart bestand genaamd "europe_countries.csv".

```python
import pandas as pd

dataframe = pd.read_csv('countries-1.csv')

# 1. Filter de dataset om alleen landen weer te geven met een bevolking van meer dan 50 miljoen
groot_bevolking = dataframe[dataframe['Population'] > 50000000]
print("\n1. Landen met een bevolking van meer dan 50 miljoen:")
print(groot_bevolking[['Country', 'Population']].to_string(index=False))

# 2. Sorteer de dataset op basis van de oppervlakte van de landen, van groot naar klein
gesorteerd_oppervlakte = dataframe.sort_values(by='Area (sq. mi.)', ascending=False)
print("\n2. Landen gesorteerd op oppervlakte (van groot naar klein):")
print(gesorteerd_oppervlakte[['Country', 'Area (sq. mi.)']].head(10).to_string(index=False))

# 3. Toon de top 10 landen met de hoogste bevolkingsdichtheid
hoogste_dichtheid = dataframe.sort_values(by='Pop. Density (per sq. mi.)', ascending=False)
print("\n3. Top 10 landen met de hoogste bevolkingsdichtheid:")
print(hoogste_dichtheid[['Country', 'Pop. Density (per sq. mi.)']].head(10).to_string(index=False))

# 4. Bereken de gemiddelde bevolking van alle landen in de dataset
gemiddelde_bevolking = dataframe['Population'].mean()
print(f"\n4. Gemiddelde bevolking van alle landen: {gemiddelde_bevolking:.2f}")

# 5. Groepeer de landen per continent en bereken het totaal aantal inwoners per continent
continent_totaal = dataframe.groupby('Region')['Population'].sum().reset_index()
print("\n5. Totaal aantal inwoners per continent:")
print(continent_totaal.to_string(index=False))

# 6. Wat is het kleinste en grootste land qua oppervlakte in de dataset?
kleinste_land = dataframe.loc[dataframe['Area (sq. mi.)'].idxmin()]
grootste_land = dataframe.loc[dataframe['Area (sq. mi.)'].idxmax()]
print("\n6. Kleinste en grootste land qua oppervlakte:")
print(f"Kleinste land: {kleinste_land['Country']} met een oppervlakte van {kleinste_land['Area (sq. mi.)']} vierkante mijl")
print(f"Grootste land: {grootste_land['Country']} met een oppervlakte van {grootste_land['Area (sq. mi.)']} vierkante mijl")

# 7. Exporteer alleen de landen uit Europa naar een apart bestand
europese_landen = dataframe[dataframe['Region'].str.contains('EUROPE', case=False)]
europese_landen.to_csv('europe_countries.csv', index=False)
print("\n7. Europese landen geëxporteerd naar 'europe_countries.csv'")
print(f"Aantal geëxporteerde landen: {len(europese_landen)}")
```
