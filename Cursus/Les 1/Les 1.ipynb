## Input & output
Conversion between types
- int --> integer number
- float --> decimal number
- str --> string = text!

```python
x = input("Give me a number: ")
print(x)
print("value of x:", x)
print(type(x))

# converting from one datatype to another
int_number = int(x)
print(int_number)
print(type(int_number))
print("value of x:", int_number/2)


float_number = float(x)
print(float_number)
print(type(float_number))
print("value of x:", float_number)
```

## Combining outputs --> concatenation
- '+' -> combines 2 strings or adds 2 numbers you have to provide a space and all parts have to be strings
- ',' -> provides an actual space and allows the use of different datatypes

```python
number = 5
print("The number is", number)
print(number + number) #10
print("The number is " + str(number))
print(str(number) + str(number)) #55

print(str(number + number)) #10
```

## Conditional statements
if-structure!
zero is even

```python
#find the sign of a specific number
#input() is always a string, therefore we have to convert the function

number = int(input("Enter a number:"))
#number = int(number)

if number > 0:
    print(number, "is positive")
elif number < 0:
    print(number, "is negative")
else:
    print(number, "is zero")


#other implementation
if number == 0:
    print(number, "is zero")
elif number > 0:
    print(number, "is positive")
else:
    print(number, "is negative")


#determine whether a number is even or odd
#zero is even

if number % 2 == 0:
    print(number, "is even")
else:
    print(number, "is odd")
```

```python
#nr 1
from datetime import datetime as dt

year = input("Enter your year of birth:")
age = dt.now().year - int(year)

if age >= 18:
    print("you are eligible to vote")
else:
    print("you are NOT eligible to vote")
```

```python
#nr 2
stringToCheck = input("Enter a string")

if not stringToCheck:
    print("The string is empty")
else:
    print("the string is not empty")


if stringToCheck == "":
    print("The string is empty")
else:
    print("the string is not empty")
```

```python
#nr 3

Charcheck = input("Give me a password")
if len(Charcheck) >= 8 and any(char.isdigit() for char in Charcheck) and any(char.isalpha() for char in Charcheck):
    print("The password is secure")
else:
    print("The password is NOT secure")
```

## Loops
- <b>for-loop</b>: loopen over een sequentie of een range, je weet exact hoeveel keer je loopt.
- **while-loop**: loopen zolang aan een bepaalde voorwaarde wordt voldaan. Geen idee hoeveel keer op voorhand

```python
#een woord is ook een sequentie --> een sequentie of opeenvolging van characters

for c in "een woord":
    print(c)

for c in "een woord":
    print(c,end = "") # --> end weglaten uit print() --> standaard \n voor nieuwe lijn

```

```python
for i in range(0,11):
    print(i)
print()
for x in range(-5,5):
    print(x)
print()
for x in range(0,11):
    print("5 *",x,"=",x*5)
```

```python
waarde = input("Geef iets in, stopt pas bij 'STOP'")
while waarde != "STOP":
    print(waarde)
    waarde = input()
```

```python
x = 5
while x<10000:
    x*=5
    print(x,end=" ")
print()

x = 5
while x<10000:
    print(x,end=" ")
    x*=5
print()
```

## Voorwaarden combineren
- **and**: beide moeten kloppen
- **or**: minstens 1 voorwaarde moet kloppen
- **not**: omgekeerde van voorwaarde

```python
x,y,z = 1,2,3
print(x<y and y<z)
print(x<y or y<z)
print(x>y or y>z)
print(x>y and y>z)
print(x<y and y>z)
print(x<y or y>z)
print(not(x<y and y>z))
print(not(x<y or y>z))
```

### Nog enkele oefeningen op loops
- Oef 1: FizzBuzz: Getallen printen van 1 tot 100, veelvoud van 3 --> Fizz, veelvoud van 5 --> Buzz, beide --> FizzBuzz

```python
for i in range(1,101):
    if i%3 == 0 and i%5 == 0:
        print("FizzBuzz")
    elif i%5 == 0:
        print("Buzz")
    elif i%3 == 0:
        print("Fizz")
    else:
        print(i)
```
